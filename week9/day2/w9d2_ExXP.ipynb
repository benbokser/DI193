```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
```

```python
## Scraping from `books.toscrape.com`

#     Extract necessary data such as plan names, features, and pricing.
#     Store and Save the Data
#     Close Selenium WebDriver

import nest_asyncio
from playwright.async_api import async_playwright
import asyncio

nest_asyncio.apply()

async def scrape_books_to_scrape():
    product_data = []
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        target_url = "http://books.toscrape.com/"
        await page.goto(target_url, timeout=90000, wait_until="domcontentloaded")

        print(f"Navigated to: {page.url}")
        print(f"Page title: {await page.title()}")

        try:
            # Wait for a prominent element that indicates the page has loaded content
            await page.wait_for_selector(".page_inner", state="visible", timeout=60000)
            print("Successfully waited for '.page_inner' element.")
        except Exception as e:
            print(f"Could not find '.page_inner' element: {e}")
            await browser.close()
            return product_data

        # --- Data Extraction Logic for books.toscrape.com ---
        # Each book is contained within an article tag with class 'product_pod'
        book_cards = await page.query_selector_all("article.product_pod")

        if not book_cards:
            print("No book cards found. Please check the website's HTML structure.")

        for card in book_cards:
            # Extract Title
            title_element = await card.query_selector("h3 a")
            title = await title_element.get_attribute('title') if title_element else 'N/A'

            # Extract Price
            price_element = await card.query_selector("p.price_color")
            price = await price_element.text_content() if price_element else 'N/A'
            price = price.strip() if price != 'N/A' else price

            # Extract Rating
            rating_element = await card.query_selector("p.star-rating")
            rating = await rating_element.get_attribute('class') if rating_element else 'N/A'
            if rating != 'N/A':
                rating = rating.replace('star-rating ', '') # Clean up the class name

            product_data.append({
                'title': title,
                'price': price,
                'rating': rating
            })

        await browser.close()
    return product_data

# Run the async function
scraped_books = asyncio.run(scrape_books_to_scrape())

# Display the scraped data
print("\nScraping books.toscrape.com finished.")
if scraped_books:
    for item in scraped_books:
        print("---------------------")
        print(f"Title: {item['title']}")
        print(f"Price: {item['price']}")
        print(f"Rating: {item['rating']}")
else:
    print("No data was scraped from books.toscrape.com.")
```

```python
import pandas as pd

# Create a Pandas DataFrame from the scraped data
df_books = pd.DataFrame(scraped_books)

print("\nScraped Data as DataFrame:")
display(df_books.head())
```

```python
# Save the DataFrame to a CSV file
csv_filename = "scraped_books.csv"
df_books.to_csv(csv_filename, index=False)
print(f"\nData saved to {csv_filename}")
```
