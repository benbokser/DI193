```python
!pip install rouge_score
import nltk
from nltk.translate.bleu_score import sentence_bleu
from rouge_score import rouge_scorer

# Ensure NLTK data is downloaded
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')

# 2. Applying BLEU and ROUGE Metrics:

# Calculate the BLEU score for the following example:
#     Reference: "Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation."
#     Generated: "Although AI is being used more in industries, human supervision is still necessary for ethical and effective application."

reference_bleu = ["Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation.".lower().split()]
generated_bleu = "Although AI is being used more in industries, human supervision is still necessary for ethical and effective application.".lower().split()

bleu_score = sentence_bleu(reference_bleu, generated_bleu)
print(f"BLEU Score: {bleu_score:.4f}")

# Calculate the ROUGE score for the following example:
#     Reference: "In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact."
#     Generated: "To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development."

scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

reference_rouge = "In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact."
generated_rouge = "To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development."

scores = scorer.score(reference_rouge, generated_rouge)

print("\nROUGE Scores:")
for key, value in scores.items():
    print(f"{key}: P={value.precision:.4f}, R={value.recall:.4f}, F={value.fmeasure:.4f}")
```

### Analysis of the limitations of BLEU and ROUGE when evaluating creative or context-sensitive text.

1.  **Lack of Semantic Understanding:** Both BLEU and ROUGE are primarily based on lexical overlap (n-gram matching). They do not understand the meaning or semantics of the text. This means they can assign low scores to perfectly valid, semantically equivalent, but lexically different sentences, which is common in creative or context-sensitive generation where paraphrasing and varied expression are expected.
2.  **Inability to Assess Fluency/Coherence:** These metrics struggle to evaluate the overall flow, grammar, and logical consistency of generated text. A generated text might have high n-gram overlap but be grammatically incorrect or nonsensical.
3.  **Sensitivity to Word Choice/Synonyms:** If a generated text uses synonyms or different phrasings that convey the same meaning as the reference, BLEU and ROUGE might penalize it heavily due to the lack of exact word matches.
4.  **Reference Dependency:** They require one or more human-written reference texts, which can be problematic for open-ended or creative tasks where multiple valid outputs are possible, and creating a comprehensive set of references is difficult or impossible.
5.  **Bias towards Shorter Texts (BLEU):** BLEU can sometimes reward shorter generated sentences that have higher precision due to fewer n-grams to match, even if they omit important information.
6.  **Difficulty with Factual Accuracy:** They do not assess factual correctness. A generated text could be factually wrong but still get a decent score if it shares n-grams with a reference that is also factually wrong or if it manages to match n-grams from a correct reference despite being incorrect in other parts.

### Improvements or Alternative Methods for Evaluating Text Generation:

1.  **Human Evaluation:** Remains the gold standard. Humans can assess subjective qualities like creativity, fluency, coherence, factual accuracy, relevance, and overall quality. This is crucial for creative or context-sensitive tasks.
2.  **Semantic Similarity Metrics (e.g., BERTScore, MoverScore):** These metrics leverage contextual embeddings from pre-trained language models (like BERT) to measure semantic similarity between generated and reference texts, rather than just lexical overlap. They can capture meaning more effectively.
3.  **Fact-checking Metrics:** Specialized metrics or models designed to verify the factual accuracy of generated content, especially for tasks like summarization or question answering.
4.  **Adversarial Evaluation:** Involving human evaluators or models to find weaknesses, biases, or undesirable outputs, pushing the model to its limits.
5.  **Task-Specific Metrics:** Developing metrics tailored to the specific goals of the generation task. For instance, in dialogue systems, metrics like turn-level engagement or task completion might be more relevant.
6.  **Reinforcement Learning from Human Feedback (RLHF):** Incorporating human preferences directly into the model's training process, guiding it towards generating more desirable and contextually appropriate outputs.
7.  **Qualitative Analysis:** Beyond scores, in-depth manual review of a subset of outputs to understand patterns of errors, strengths, and emergent behaviors.

```python
# 3. Perplexity Analysis:
#
# *   **Compare the perplexity of the two language models based on the probability assigned to a word:**
#     *   Model A: Assigns 0.8 probability to “mitigation.”
#     *   Model B: Assigns 0.4 probability to “mitigation.”
#
#     **Calculation and Comparison:**
#     Perplexity is inversely related to probability (Perplexity = 1 / Probability).
#     *   **Model A Perplexity:** 1 / 0.8 = 1.25
#     *   **Model B Perplexity:** 1 / 0.4 = 2.5
#
#     **Determine which model has lower perplexity and explain why:**
#     Model A has a lower perplexity (1.25) compared to Model B (2.5). This is because Model A assigns a higher probability (0.8) to the word “mitigation.” A lower perplexity indicates that the model is more confident and 'less surprised' by the word, meaning it predicts the word with higher accuracy within its context.
#
# *   **Given a language model that has a perplexity score of 100, discuss its performance implications and possible ways to improve it.**
#
#     **Performance Implications of Perplexity Score of 100:**
#     A perplexity score of 100 is generally considered high, especially for typical language modeling tasks where lower scores are desired. It implies that the model is, on average, as uncertain about the next word as if it had to choose uniformly from 100 equally likely words. This high uncertainty suggests:
#     *   **Poor Predictive Performance:** The model likely struggles to predict the next word accurately in a sequence.
#     *   **Low Fluency/Coherence:** Generated text might be less natural, fluent, or coherent, frequently making less probable word choices.
#     *   **Limited Understanding:** The model may not have a strong grasp of linguistic patterns, grammar, or context within the training data.
#     *   **High 'Surprise':** For common sequences, the model is often 'surprised' (assigns low probability), indicating a mismatch between its internal representation and the actual language data.
#
#     **Possible Ways to Improve a Language Model with Perplexity Score of 100:**
#     1.  **Increase Training Data Size and Quality:** More diverse and relevant training data can help the model learn better language patterns. Cleaning noisy data is also crucial.
#     2.  **Improve Model Architecture:** Experiment with more complex or suitable architectures (e.g., larger transformer models, better attention mechanisms) that can capture long-range dependencies more effectively.
#     3.  **Hyperparameter Tuning:** Optimize learning rate, batch size, number of layers, hidden units, etc., to ensure the model trains efficiently and effectively.
#     4.  **Regularization Techniques:** Apply dropout, L1/L2 regularization, or early stopping to prevent overfitting, which can lead to high perplexity on unseen data.
#     5.  **Pre-training and Fine-tuning:** Utilize pre-trained models on vast text corpora and then fine-tune them on a specific domain or task-specific dataset. This leverages general language understanding.
#     6.  **Better Tokenization:** Ensure an appropriate tokenization strategy that aligns with the language and task requirements.
#     7.  **Ensemble Methods:** Combine predictions from multiple models to achieve a more robust and accurate prediction.
#     8.  **Contextual Window Expansion:** Allow the model to consider a larger context window when predicting the next word.
```

```python
# 6. Comparative Analysis of Evaluation Methods:
#
#     Choose an NLP task (e.g., machine translation, text summarization, question answering).
#     Compare and contrast at least three different evaluation metrics (BLEU, ROUGE, BERTScore, Perplexity, Human Evaluation, etc.).
#     Discuss which metric is most appropriate for the chosen task and why.

# **Chosen NLP Task: Question Answering (QA)**
#
# Question Answering is an NLP task where a model receives a natural language question and a context (e.g., a document or passage) and is expected to provide a precise answer. QA can be broadly categorized into extractive (where the answer is a span of text directly from the context) and abstractive (where the answer is generated based on understanding the context, not necessarily a direct copy).
#
# **Comparison of Evaluation Metrics:**
#
# 1.  **Exact Match (EM)**
#     *   **Description:** EM measures whether the system's generated answer string is character-for-character identical to any of the reference answers. It's a binary metric (1 if exact, 0 otherwise).
#     *   **Strengths for QA:** Very clear and objective. If an exact answer is required (e.g., for specific entities like names, dates, or numbers), EM provides a strong indicator of correctness. It is simple to compute and interpret.
#     *   **Weaknesses for QA:** Extremely strict. It penalizes answers that are semantically correct but syntactically different (e.g., "USA" vs. "United States of America"). It doesn't give partial credit and can be unforgiving with minor punctuation or capitalization differences. Not suitable for abstractive QA where paraphrasing is common.
#
# 2.  **F1 Score**
#     *   **Description:** F1 score is a token-based metric that calculates the harmonic mean of precision and recall between the predicted answer and the ground truth answer. Precision is the proportion of words in the prediction that also appear in the ground truth, and recall is the proportion of words in the ground truth that appear in the prediction.
#     *   **Strengths for QA:** Less strict than EM, allowing for partial credit. It captures semantic overlap better by considering shared tokens, making it more robust to slight variations in wording. It's widely used for extractive QA and provides a good balance between precision and recall.
#     *   **Weaknesses for QA:** Still token-based, so it might not fully capture semantic equivalence if different words are used to convey the same meaning. It doesn't consider grammatical correctness, fluency, or overall coherence, especially for longer answers.
#
# 3.  **Human Evaluation**
#     *   **Description:** Involves human annotators assessing the quality of answers based on criteria such as correctness, fluency, completeness, conciseness, and relevance. Ratings are often on a Likert scale or comparative judgments.
#     *   **Strengths for QA:** The gold standard for assessing overall quality. Humans can discern semantic nuances, factual accuracy, fluency, and appropriateness that automated metrics miss. Essential for evaluating abstractive QA where answers can vary widely and for identifying subtle errors or biases.
#     *   **Weaknesses for QA:** Expensive, time-consuming, and can be subjective, requiring careful annotation guidelines and inter-annotator agreement checks. Difficult to scale for large datasets or frequent evaluations.
#
# 4.  **BERTScore**
#     *   **Description:** BERTScore leverages contextual embeddings from pre-trained language models (like BERT) to measure the semantic similarity between the generated answer and the reference answer. It computes a soft F1-like score based on token embeddings rather than exact token matches.
#     *   **Strengths for QA:** Addresses the limitations of EM and F1 by capturing semantic equivalence even when different words are used. It's particularly powerful for abstractive QA, where answers are often paraphrased. It correlates better with human judgment than token-overlap metrics.
#     *   **Weaknesses for QA:** Computationally more intensive than EM or F1. Requires a pre-trained language model. While better at semantics, it might still struggle with factual accuracy if the model misunderstands the context or generates plausible but incorrect information.
#
# **Most Appropriate Metric for Question Answering:**
#
# For **extractive Question Answering**, a combination of **F1 Score** and **Exact Match (EM)** is often most appropriate. EM provides a strict measure of perfect answers, while F1 offers a more nuanced view by giving partial credit for semantically similar but not identical answers. This pair gives a comprehensive view of how well the model extracts the correct span.
#
# For **abstractive Question Answering**, which involves generating novel text, **BERTScore** or similar embedding-based metrics are highly recommended as they capture semantic similarity more effectively than token-overlap metrics. However, for a complete and robust evaluation, **Human Evaluation** remains indispensable, especially for critical applications where factual correctness, fluency, and safety are paramount. Human evaluation can identify subtle errors, logical inconsistencies, or biases that even advanced automated metrics might miss. Therefore, a multi-faceted approach combining automated metrics like F1/EM/BERTScore with strategic human evaluation offers the most comprehensive assessment for QA systems.
```
