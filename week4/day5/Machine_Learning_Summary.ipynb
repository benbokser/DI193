# About Machine Learning

Machine Learning (ML) is a method in which digital systems learn from data to make decisions or predictions, without being explicitly programmed for each task.  
Machine Learning is important for data analysts because it allows them to move beyond descriptive work (analysis and visualization of what happened in the past) into predictive and prescriptive analysis. It is especially helpful with large and complex datasets, enabling automation of repetitive tasks, revealing hidden correlations that human observers might miss, and providing complex and scalable insights.





# Key ML Applications Across Industries
ML has been very useful in many industries. Here are three examples:
•	Finance: Financial Institutions use ML for fraud detection. ML models are used to analyze millions of transactions in real time, identifying anomalies (such as unusual locations or spending patterns) to flag and block suspicious activity instantly.
•	Healthcare: ML is used in imaging and diagnostics. Models trained on large sets of X-rays or MRIs are used to assist radiologists in identifying issues of concern with precision.
•	E-commerce: retailers like Amazon, Netflix, and Spotify use past browsing and consumption history combined with similar user behavior to suggest products, movies, or music users are likely to enjoy, increasing customer satisfaction, engagement, and revenue.






# Types of Machine Learning
There are three types of Machine Learning:
# Supervised Learning
In this type, the learning is “supervised” in that the data is already labeled with the correct output. The ML model develops a function to predict the output for new data.
•	Scenario: Loan Default. A model is given data on loans and whether they defaulted or not. Once trained, it can be given data on a new loan and predict whether it will default.
# Unsupervised Learning
In this type, the model is trained on unlabeled data lacking correct outputs. Instead of learning to predict an output, the model finds inherent structures, patterns, or groupings within the data on its own.
•	Scenario: Playlist Creation. A music streaming service like Spotify uses ML to group songs into clusters based on their likelihood to be played and liked by the same users. These clusters are then presented to users as suggested playlists.
#Reinforcement Learning
In this type, an ML Agent learns to make decisions by performing actions in an environment to maximize a Cumulative Reward. It learns through trial and error, receiving feedback in the form of rewards or penalties.
•	Scenario: LLM Training. In part of the process of training Artificial Intelligence Large Language Models (LLMs), LLM responses are provided feedback by human reviewers on metrics such as response quality and harmfulness. Over time it adjusts it learns from this feedback to provide responses that perform better on these metrics.








# Developing a Model
1.	Feature Selection
Feature selection involves identifying the most relevant variables (features) for a model’s predictive power and choosing those variables for the model to use in its training. Removing redundant or irrelevant data reduces “noise” in a model, making it faster to train and more accurate.
2.	Model Selection
Model selection involves selecting the algorithm to be used, considering the nature of the data (size, type, and quality) and the goal. Often, analysts test several models and compare their baseline performances.
3.	Model Evaluation
After a model is trained, it is evaluated on data it hasn’t seen (the “Test Set”). Evaluation involves using metrics to determine the quality of the model’s predictions. Basic metrics include accuracy, precision, and recall. If a model that performs well on training data performs poorly on test data, it is likely “overfitting”—like a student who has memorized the answers to a practice test rather than learning the underlying content, it has excessively learned to predict based on the noise rather than learning underlying patterns. This requires analysts to go back and refine the features or the model (or choose a different model).
